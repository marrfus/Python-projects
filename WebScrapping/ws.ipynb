```python
# requests, lxml, beautifulsoup4
import requests
from bs4 import BeautifulSoup
import csv

#date = input("Please enter a date in the following formate YYYY-MM-TT:")
date = "2026-03-24"
page = requests.get(f"https://www.bbc.com/sport/football/scores-fixtures/{date}")

def main(page):
    src = page.content
    soup = BeautifulSoup(src, "lxml")
    matches_details=[]
 
    championship = soup.find_all("div",{"class":"ssrcss-7k0bq5-HeaderWrapper"})
    
    def get_match_info(data):
        champioship_title = data.contents[0].find("a").text.strip()
        all_matches = data.contents[0].find_all("li")
        #print(champioship_title)
        #print(len(all_matches))
        for i in range(len(all_matches)):
            team_home_data = all_matches[i].find("div", {"class":"ssrcss-bon2fo-WithInlineFallback-TeamHome"})
            #team_home_div = team_home_data.find("div",{"class","ssrcss-p49l04-TeamNameWrapper"})
            team_home = team_home_data.find("span",{"class":"ssrcss-c8w0oz-MobileValue"}).text.strip()

            team_away_data = all_matches[i].find("div", {"class":"ssrcss-nvj22c-WithInlineFallback-TeamAway"})
            team_away = team_away_data.find("span",{"class":"ssrcss-c8w0oz-MobileValue"}).text.strip()
            #print(f"{team_home} : {team_away}")
            
            # get scores
            home_score = all_matches[i].find("div",{"class":"ssrcss-qsbptj-HomeScore"}).text.strip()
            away_score = all_matches[i].find("div",{"class":"ssrcss-fri5a2-AwayScore"}).text.strip()
            #print(f"{home_score} : {away_score}")

            matches_details.append({"championship":champioship_title, "Team Home":team_home, "Team Away":team_away,
                                    "Home Score":home_score,"Away Score":away_score})
            

        #print(all_matches)
    for i in range(len(championship)):
        get_match_info(championship[i])
    
    keys = matches_details[0].keys()
    
    with open(f"matches_details_{date}.csv", "w") as f:
        dict_writer = csv.DictWriter(f, keys)
        dict_writer.writeheader()
        dict_writer.writerows(matches_details)
        print("file created!")

main(page)
```
