```python
import requests
from bs4 import BeautifulSoup
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
```

```python

def main(page):
    src = page.content
    soup = BeautifulSoup(src, "lxml")
    titels = soup.select("h2[data-service-review-title-typography]")
    bewertungen=[]
    for t in titels:
        bewertungen.append(t.text.strip())
        print(t.text.strip())
    
    return bewertungen
    
```

```python
headers = {
     "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)",
    "Accept-Language": "en-US,en;q=0.9",
    "Accept": "text/html",
    "Connection": "keep-alive"
}
source = requests.get("https://www.trustpilot.com/review/www.enjoydus.com", headers=headers)
bewertungen = main(source)
```

```python
nltk.download("vader_lexicon")   #einmalige download
```

```python
sia = SentimentIntensityAnalyzer()  #für Englisch Texte verwendet!!!

for text in bewertungen:
    print("Text:", text)

    score = sia.polarity_scores(text)

    for kategorie, wert in score.items():
        print(f" - {kategorie}: {wert}")
    print("-"*30)
```
