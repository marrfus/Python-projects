```python
# spaCy installieren + Modell laden
# pip install spacy 
# python -m spacy download de_core_news_sm

import spacy 
```

Vervollständige den Code, um nur die Entitäten auszugeben, die entweder ein Ort oder eine 
Organisation sind. 

```python
# 1. Das deutsche Modell laden 
nlp = spacy.load("de_core_news_sm") 
 
text = """ 
Gestern traf sich der Vorstand der Siemens AG in München. 
Dort wurde beschlossen, dass die Zusammenarbeit mit Google in Berlin 
weiter ausgebaut werden soll. Angela Merkel wurde als Gast ebenfalls 
in der bayerischen Landeshauptstadt erwartet. 
""" 
```

```python
# 2. Den Text verarbeiten

doc = nlp(text) 
#verarbeitet den Text und wandelt ihn in ein Doc-Objekt um,
# das viele Informationen enthält
 
def extrahiere_infos(spacy_doc): 
   print(f"{'Textstelle':<20} | {'Typ':<10} | {'Erklärung'}") 
   print("-" * 50)
   
   for ent in spacy_doc.ents: 
       # DEINE AUFGABE HIER: 
       # Filtere so, dass nur 'ORG' (Organisationen) und 'LOC' (Orte)
       # ausgegeben werden. 'PER' (Personen) soll ignoriert werden. 
       if ent.label_ =='ORG'or ent.label_ =='LOC':
            print(f"    {ent.text:<20}  {ent.label_:<10}{spacy.explain(ent.label_)}")
            
       
       # Tipp: ent.label_ enthält den Typ (z.B. "ORG") 
       # Tipp: spacy.explain(ent.label_) gibt eine Erklärung zum Typ 
       
       pass # Ersetze dies durch deine Logik 
```

```python
extrahiere_infos(doc)
```

```python
#Visual
from spacy import displacy
displacy.render(doc, style="ent")
```
