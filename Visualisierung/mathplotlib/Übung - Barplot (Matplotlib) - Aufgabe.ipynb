```python
# benötigte bibs importieren

import matplotlib 
import matplotlib.pyplot as plt 
import matplotlib.gridspec as gridspec
import seaborn as sns # um einen Beispieldatensatz zu haben
import pandas as pd
import numpy as np
```

```python
# datensatz aus seaborn laden

df_tips = sns.load_dataset('tips')
df_tips 
```

Aufgabe 1: (cut-Funktion)

Für die Tischgröße (size) sollen Kategorien angelegt werden.

Dabei sollen einzelne Personen als Single deklariert werden, 2er Leute als Duo und 3-6 soll als Gruppe ausgewiesen werden.

Ziel: neue Spalte mit Single, Duo, Gruppe


```python
# platz für deinen Code 
categories = ["Single", "Duo", "Gruppe"]
df_tips["tishgroesse"] = pd.cut(df_tips["size"], bins=[0,1,2,6], labels=categories)
df_tips.head()




```
